Let's solve the linear system
$$
\begin{align}
2a + 3b + 4d &= 0,\\
a + 9c - 3d &= 0
\end{align}
$$ modulo the prime number 97

```python
equation1 = [["a",2], ["b",3], ["d",4]]    # 2a + 3b + 4d = 0
equation2 = [["a",1], ["c",9], ["d", -3]]  # a + 9c - 3d = 0

equation1 = [[(1,1),2], [(2,2),3], [(4,4),4]]    # 2a + 3b + 4d = 0
equation2 = [[(1,1),1], [(3,3),9], [(4,4), -3]]  # a + 9c - 3d = 0
```

```python
equations = [equation1, equation2]
variables = ["a", "b", "c", "d"]
variables = [(1,1), (2,2), (3,3), (4,4)]
modulus = 97
```

```python
print(equations)
```

```python
pyfeyngym.solve_eqs_modulo(equations, variables, modulus)
```

The above solution is:
$$
\begin{align}
b &= -91c + 29d,\\
a &= 88c - 94d
\end{align}
$$
Notice that the solver always attempt to solve for the variables that appear near the beginning of the list in terms of variables that apper near the end of the list.
When solving large systems, you'll see the "current row" update in interval of 100 equations, 1, 101, 201...

You can also use simple integer labels 1, 2 ... to indicate variable 1, variable 2 ... , instead of naming the variables with strings:

```python
equation1_a = [[1,2], [2,3], [4,4]]    # 2var[1] + 3var[2] + 4var[4] = 0
equation2_a = [[1,1], [3,9], [4, -3]]  # var[1] + 9var[3] - 3var[4] = 0
equations_a = [equation1_a, equation2_a]
variables_a = [1, 2, 3, 4]
modulus = 97
pyfeyngym.solve_eqs_modulo(equations_a, variables_a, modulus)
```
