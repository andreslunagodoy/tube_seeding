```python
# This notebook demonstrates the one-loop bubble RL environment without training
```

```python
from pyfeyngym import IBPEnv, visualize_board
import gymnasium as gym
import numpy as np
```

```python
# Create an environment with a randomly selected target integral
env = IBPEnv(random_target = True)
```

```python
# Equivalently, you could write
env = gym.make("pyfeyngym-v0", random_target = True)
env = env.unwrapped # Access the raw environment which has more methods than the standard gymnasium interface
```

```python
# Set tk backend for matplotlib if possible
from IPython import get_ipython
ip = get_ipython()
try:
    ip.run_line_magic("matplotlib", "tk")
except Exception:
    ip.run_line_magic("matplotlib", "inline")
```

```python
# Currently, visualizing the environemnt shows just an empty board with the target integral marked as a red square
fig, ax, anim = env.visualize()
```

```python
# This is the list of all possible actions, as (x coordinate, y coordinate, action type).
# x and y coordinates are the powers of the two propagators of the bubble Feynman diagram
# action type 1 and 2 are the two types of IBP equations, and action type 0 corresponds to
# eliminating the integral in Gaussian elmination
[np.array(a) for a in env.all_actions_2d] # np.array converts Julia arrays to numpy arrays for display here
```

```python
# let's look at the valid actions at this step, given as numerical indices with respect to the above list
env.define_valid_actions()
```

```python
# Take an arbitrary action, e.g. 92
observation, reward, terminated, truncated, info = env.step(92)
```

```python
# observation is a 2D image with resolution 9*9 with 4 channels. The first three channels simply
# indicate whether the three types of actions have been taken for the particular cell, and the 4th
# channel is 0 for most cells and 1 for the cell of the target integral.
observation
```

```python
# info contains the action mask, a boolean array indicating which ones in the list of all actions are valid
info
```

```python
# visualizing the environment now shows one seed integral / IBP operator used
fig, ax, anim = env.visualize()
```

```python
# Let take an action, again the first one in the list of valid actions
observation, reward, terminated, truncated, info = env.step(env.define_valid_actions()[0])
```
